```python
import sqlite3
import pandas as pd

# built into Python
# for readable output

con = sqlite3.connect('C:\\Users\\lenovo\\airtel-telecom-analytics\\data\\raw\\airtel.db') # two dots: notebooks/

def q(sql):
    return pd.read_sql(sql, con)

# the same helper as
# the IPL project
```

```python
def run_sql_file(path):                         # one .sql file
    with open(path, 'r') as f:                  # read only
        script = f.read()                       # whole file
    con.executescript(script)                   # every statement
    con.commit()                                # save to the .db
    print('ran', path)                          # show what ran
    
run_sql_file(r'C:\\Users\\Lenovo\\airtel-telecom-analytics\\sql\\practice\\00_test.sql')              # apply the file
q('SELECT COUNT(*) AS n FROM customers;')       # check it worke
```
